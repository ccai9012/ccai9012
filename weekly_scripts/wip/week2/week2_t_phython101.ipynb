## Project: build a tiny text adventure

We will use Python basics to make a small text adventure. Each section adds to the same game: print the welcome, store a character and inventory, compare choices, branch with `if`, and wrap a repeated action in a function. Run each section in order; the final output is a playable short scene in the notebook. No external data or credentials are needed.


## 1. Start the adventure: print a welcome

Run this cell to display a welcome message. `print` shows the content inside its parentheses.

```python
print("Welcome, adventurer!")  # This prints text to the screen
```

## 2. Create your hero: variables

Change the name and run the cell again. Notice how later text uses this variable.

```python
# A variable is a label that stores a value.
hero_name = "Amina"
hero_health = 10
print(f"{hero_name} starts with {hero_health} health points.")
```

## 3. Store hero stats and inventory: data types

See how text, numbers, and a list store character details. The output shows the values currently stored.

```python
# Store different kinds of information about the same game character.
hero_level = 1          # integer: completed levels
potion_strength = 3.14  # float: health restored by a potion
inventory_item = "map" # string: an item name
has_compass = True      # boolean: a yes/no state

print("Level:", hero_level, type(hero_level))
print("Potion strength:", potion_strength, type(potion_strength))
print("Inventory item:", inventory_item, type(inventory_item))
print("Has compass:", has_compass, type(has_compass))
```

### Step A — Calculate a health potion bonus

Calculate the health after using a potion. Change a number, run the cell again, and check the result.

```python
hero_health = 7
potion_bonus = 3

print("Health after potion:", hero_health + potion_bonus)
print("Health lost in a small battle:", hero_health - potion_bonus)
print("Two potions restore:", potion_bonus * 2)
print("Split the reward between two players:", 10 / 2)
print("Three potions restore:", potion_bonus ** 3)
print("Health points left after a 5-point hit:", hero_health % 5)
```

### Step B — Build the story message

Combine character details into a story message. Notice how the variables inside the braces are filled in.

```python
hero_name = "Amina"
item = "a map"
greeting = "Welcome, " + hero_name + "! You found " + item + "."
print(greeting)

# An f-string inserts current game values into the story.
hero_health = 10
message = f"{hero_name} has {hero_health} health points."
print(message)
```

### Step C — Check the hero stats

Run the check and confirm that the character state matches the choice from the previous step.

```python
hero_health = 5
safe_health = 10

# Comparisons return True or False and can control a game decision.
print("At full health?", hero_health == safe_health)
print("Needs a potion?", hero_health != safe_health)
print("Health is low?", hero_health < safe_health)
print("Health is safe?", hero_health >= safe_health)
```

## 4. Choose a path: if/else

Change the health value and see which branch runs when the condition is true or false.

```python
# The game responds differently depending on the hero's health.
hero_health = 5

if hero_health <= 3:
    print("Use a health potion before entering the cave.")
elif hero_health < 10:
    print("You can enter, but be careful.")
else:
    print("You are ready for the adventure!")
```

## 5. Reuse a game action: functions

Call the function to perform a repeatable action. It takes the character state as input and returns the updated result.

```python
# A function packages a repeatable action.
def greet_hero(name):
    return f"Welcome to the quest, {name}!"

def drink_potion(current_health, potion_bonus):
    return current_health + potion_bonus

hero_name = "Amina"
hero_health = 7
print(greet_hero(hero_name))
hero_health = drink_potion(hero_health, 3)
print(f"After drinking a potion, {hero_name} has {hero_health} health points.")
```
