## Part 2 — Parameter deep dive: temperature

For one next-token step, the model assigns scores to possible continuations. Temperature changes how sharply those scores favour the leading candidate before sampling: lower settings concentrate choices; higher settings give alternatives more weight. The figure uses a continuation of the robot-painting story from the code below, but its percentages are invented teaching values.

In the live experiment, keep the story prompt fixed and compare three responses at each temperature. At `0.0`, the hosted service may use its own near-deterministic behaviour; repeated outputs are still worth checking. Temperature affects sampling, not whether the story is accurate or useful. See the [Transformers generation reference](https://huggingface.co/docs/transformers/main/main_classes/text_generation) for the sampling parameter.

## Part 3 — Role of context: zero-shot, one-shot, few-shot

Keep the **same target sentence and requested labels** in all three calls. Zero-shot gives the task alone; one-shot adds one labelled example; few-shot adds two. Predict whether the extra examples change the output label or format. A change may reflect pattern-following rather than better judgment.

### Add an instruction or examples

```python
# Zero-shot: no labelled example.
target = "The new trees will shade the bus stop."
prompt_zero = f"Label the sentence SUPPORT or CONCERN. Reply with one label only.\nSentence: {target}\nLabel:"
llm_utils.ask_llm(prompt_zero, llm)
```

```python
# One-shot: one example before the same target.
prompt_one = f"Label the sentence SUPPORT or CONCERN. Reply with one label only.\nSentence: The road is too narrow for a cycle lane.\nLabel: CONCERN\nSentence: {target}\nLabel:"
llm_utils.ask_llm(prompt_one, llm)
```

```python
# Few-shot: two examples before the same target.
prompt_few = f"Label the sentence SUPPORT or CONCERN. Reply with one label only.\nSentence: The road is too narrow for a cycle lane.\nLabel: CONCERN\nSentence: A shaded path would make walking more comfortable.\nLabel: SUPPORT\nSentence: {target}\nLabel:"
llm_utils.ask_llm(prompt_few, llm)
```

### Read the comparison

Record the three labels and whether each response followed the one-label instruction. What stayed fixed, and what changed? The two examples above include one concern and one supportive statement. If every example favoured one label, the framing itself might steer the answer; try that as a separate experiment and discuss why it would be a biased prompt design.
