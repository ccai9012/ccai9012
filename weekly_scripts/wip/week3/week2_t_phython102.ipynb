# Python 102: Classes and Methods

In this tutorial, we will explore **methods** and **classes**, which are the foundation of object-oriented programming (OOP) in Python.

We'll cover:
1. How to define a class
2. How to create objects (instances)
3. How to define and use methods
4. How to work with class and instance attributes

## 1. Defining a Simple Class

Define the details and actions that each person will have. This step does not create a specific person yet.

```python
# Define a class called Person
class Person:
    # Class attribute: shared by all instances
    species = "human"

    # Constructor method: called when creating a new object
    def __init__(self, name, age):
        self.name = name  # instance attribute
        self.age = age    # instance attribute

    # Method to describe the person
    def description(self):
        return f"{self.name} is {self.age} years old"

    # Method to make the person "speak"
    def speak(self, sound):
        return f"{self.name} says {sound}"
```

## 2. Creating Objects (Instances)

Create two people from the same class. Compare their names and ages to see that each object stores its own details.

```python
# Create two Person objects
person1 = Person("Amina", 3)
person2 = Person("Kai", 5)

# Access instance attributes
print(person1.name)  # Amina
print(person2.age)   # 5

# Call methods
print(person1.description())  # Amina is 3 years old
print(person2.speak("Woof!")) # Kai says Woof!
```

## 3. Modifying Object Attributes

Change the age of one object, then inspect the other to see which object was affected.

```python
# Update person1's age
person1.age = 4
print(person1.description())  # Amina is 4 years old

# Add a new attribute dynamically
person1.color = "brown"
print(f"{person1.name} is {person1.color}")  # Amina is brown
```

## 4. Class Attributes

See how multiple objects share a class attribute. Try changing it and check the value each object sees.

```python
# Access class attribute
print(person1.species)  # human
print(person2.species)  # human

# Change class attribute for all instances
Person.species = "Person"
print(person1.species)  # Person
print(person2.species)  # Person
```

## 5. Summary

- A **class** is a blueprint for objects.
- An **object** (or instance) is a specific instance of a class.
- **Attributes** store data about the object.
- **Methods** define actions the object can perform.
- `self` refers to the specific object instance.
- Class attributes are shared by all objects, while instance attributes are unique to each object.
